```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load environment variables
from dotenv import load_dotenv
load_dotenv()

bucket_name = os.getenv('BUCKET_NAME')
print(f"Using bucket: {bucket_name}")
```

## Load Data from S3

```python
# Read CSV from S3
s3_path = f's3://{bucket_name}/data/raw/machines.csv'
df = pd.read_csv(s3_path)

print(f"Loaded {len(df):,} rows")
df.head()
```

## Basic Data Inspection

```python
# Data types and info
df.info()
```

```python
# Summary statistics
df.describe()
```

## Explore Temperature Distribution

```python
# Temperature histogram
plt.figure(figsize=(10, 6))
plt.hist(df['temperature'], bins=50, edgecolor='black')
plt.axvline(x=80, color='red', linestyle='--', label='Overheat threshold (80°C)')
plt.xlabel('Temperature (°C)')
plt.ylabel('Frequency')
plt.title('Machine Temperature Distribution')
plt.legend()
plt.show()
```

## Calculate Overheating Percentage

```python
# Percentage of overheating events
overheat_pct = (df['temperature'] > 80).mean() * 100
print(f"Overheating events: {overheat_pct:.2f}%")
```

## Explore by Machine

```python
# Temperature by machine
plt.figure(figsize=(10, 6))
df.boxplot(column='temperature', by='machine_id')
plt.axhline(y=80, color='red', linestyle='--', label='Overheat threshold')
plt.ylabel('Temperature (°C)')
plt.title('Temperature Distribution by Machine')
plt.suptitle('')
plt.show()
```
